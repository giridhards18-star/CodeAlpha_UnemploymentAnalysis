```python
import pandas as pd

df = pd.read_csv("../data/cleaned_unemployment.csv")

df.head()
```

```python
monthly_trend = df.groupby('Date')['Estimated Unemployment Rate (%)'].mean()

monthly_trend.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,6))

plt.plot(monthly_trend.index, monthly_trend.values, marker='o')

plt.title('Overall Unemployment Rate Trend')
plt.xlabel('Date')
plt.ylabel('Average Unemployment Rate (%)')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
pre_covid = df[df['Date'] < '2020-03-01']

covid_period = df[
    (df['Date'] >= '2020-03-01') &
    (df['Date'] <= '2020-06-30')
]

print("Before COVID average:", pre_covid['Estimated Unemployment Rate (%)'].mean())
print("COVID period average:", covid_period['Estimated Unemployment Rate (%)'].mean())
```

```python
covid_comparison = pd.Series({
    'Before COVID': pre_covid['Estimated Unemployment Rate (%)'].mean(),
    'COVID Period': covid_period['Estimated Unemployment Rate (%)'].mean()
})

covid_comparison
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    covid_comparison.index,
    covid_comparison.values
)

plt.title('Impact of COVID-19 on Unemployment')
plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')

plt.tight_layout()
plt.show()
```

```python
# Calculate the periods again
pre_covid = df[df['Date'] < '2020-03-01']

covid_period = df[
    (df['Date'] >= '2020-03-01') &
    (df['Date'] <= '2020-06-30')
]

# Calculate average unemployment rate for each period
covid_comparison = pd.Series({
    'Before COVID': pre_covid['Estimated Unemployment Rate (%)'].mean(),
    'COVID Period': covid_period['Estimated Unemployment Rate (%)'].mean()
})

# Create the graph
plt.figure(figsize=(8, 5))

plt.bar(covid_comparison.index, covid_comparison.values)

plt.title('Impact of COVID-19 on Unemployment')
plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')

plt.tight_layout()

# Save the graph
plt.savefig('../visualization/covid_impact.png')

plt.show()
```

```python
# Calculate average unemployment rate for each region
region_avg = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean()

# Sort regions from highest to lowest
region_avg = region_avg.sort_values(ascending=False)

# Select the top 10 regions
top_regions = region_avg.head(10)

# Create the graph
plt.figure(figsize=(10, 6))

plt.bar(top_regions.index, top_regions.values)

plt.title('Top 10 Regions by Average Unemployment Rate')
plt.xlabel('Region')
plt.ylabel('Average Unemployment Rate (%)')

plt.xticks(rotation=45)

plt.tight_layout()

# Save the graph
plt.savefig('../visualization/regional_analysis.png')

plt.show()
```

```python
# Calculate average unemployment rate for each month
monthly_avg = df.groupby('Month')['Estimated Unemployment Rate (%)'].mean()

# Create the graph
plt.figure(figsize=(10, 6))

plt.plot(monthly_avg.index, monthly_avg.values, marker='o')

plt.title('Average Unemployment Rate by Month')
plt.xlabel('Month')
plt.ylabel('Average Unemployment Rate (%)')

plt.xticks(
    range(1, 13),
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
     'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
)

plt.tight_layout()

# Save the graph
plt.savefig('../visualization/seasonal_analysis.png')

plt.show()
```

```python
# Calculate average unemployment rate for each area
area_avg = df.groupby('Area')['Estimated Unemployment Rate (%)'].mean()

# Create the graph
plt.figure(figsize=(8, 5))

plt.bar(area_avg.index, area_avg.values)

plt.title('Average Unemployment Rate: Rural vs Urban')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')

plt.tight_layout()

# Save the graph
plt.savefig('../visualization/rural_urban_analysis.png')

plt.show()
```

## Conclusion

The unemployment analysis provides important insights into unemployment trends across India.

The overall average unemployment rate in the cleaned dataset was 11.79%, while the median unemployment rate was 8.35%. The unemployment rate varied considerably across observations, ranging from 0% to 76.74%.

The analysis shows a substantial increase in unemployment during the COVID-19 period. The average unemployment rate increased from 9.51% before COVID-19 to 17.77% during March–June 2020, representing an increase of 8.26 percentage points.

Monthly analysis showed that April had the highest average unemployment rate at 23.64%, while July had the lowest at 9.03%. The April spike coincides with the COVID-19 period and therefore should not be considered purely seasonal.

There was also significant regional variation. Tripura recorded the highest average unemployment rate among the regions analyzed, with an average of 28.35%.

Overall, the analysis demonstrates significant temporal and regional differences in unemployment and highlights the strong association between the COVID-19 period and increased unemployment. These findings can help policymakers identify periods and regions that may require greater employment support and economic intervention.
